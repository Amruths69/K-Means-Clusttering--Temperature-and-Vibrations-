```python
#importing libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
```

```python
df=pd.DataFrame({
    "Temp":[30,50,67,45,32,32,34,42,21,12,44,55,66],
    "Vib":[1.2,1.3,1.4,1.5,1.6,2.4,2.6,2.8,40,4.2,4.4,4.5,5.5]
})
```

```python
df
```

```python
plt.scatter(x="Temp",y="Vib",data=df)
plt.xlabel("Temp")
plt.ylabel("Vib")
plt.title("Temp vs Vib")
```

```python
#storing value of temp and vib in x as 2d
x=df[["Temp","Vib"]]
x
```

```python
#seprating into groups & training model using fit
model = KMeans(n_clusters=3,random_state=42)
model.fit(x)
```

```python
#using keyword to create a new column with groupin 
df["Cluster"]=model.labels_
df
```

```python
#using sns and hue we are seprating the data points into three seperate groups!
sns.scatterplot(x="Temp",y="Vib",data=df,hue="Cluster")
plt.xlabel("Temp")
plt.ylabel("Vib")
plt.title("KMeans Clusttering")
```

Now craeting a center for each group

```python
#Creating center point [:,0]-> take first column marker->symbol you want
centroids=model.cluster_centers_
sns.scatterplot(x="Temp",y="Vib",data=df,hue="Cluster")
plt.scatter(centroids[:,0],centroids[:,1],marker="^",s=200,color="blue")
plt.xlabel("Temp")
plt.ylabel("Vib")
plt.title("KMeans Clustering with Centroids")
```
